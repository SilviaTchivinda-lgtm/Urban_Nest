```python
# import necessary libraries
import pandas as pd
```

```python
# load the dataset
urbannest_df=pd.read_csv('./data/UrbanNest_Retail_Dirty_Customer_Order_Dataset.csv')
```

```python
# clean the column names
urbannest_df.info()
```

```python
# convert the order_date column to datetime format
urbannest_df['order_date'] = pd.to_datetime(urbannest_df['order_date'], errors='coerce')
```

```python
# order_date column to datetime format
urbannest_df.info()
```

```python
# sample of the dataset
urbannest_df.head()
```

```python
#change gender values to Male and Female

gender = urbannest_df["gender"].astype("string").str.strip().str.casefold()
urbannest_df["gender"] = gender.replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
})
```

```python
#print the unique values in the gender column
urbannest_df["gender"]
```

```python
# replace the values in the age_group column with a more general representation of the age group

age_group = urbannest_df["age_group"].astype("string").str.strip().str.casefold()
age_group = age_group.replace({"<na>": pd.NA, "nan": pd.NA})
age_group = age_group.str.replace(r"^(\d+)-(\d+)$", r"\1+", regex=True)
urbannest_df["age_group"] = age_group
```

```python
# clean age_group column
urbannest_df["age_group"]
```

```python
# clean the city column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, convert the cleaned city names to title case.

city = urbannest_df["city"].astype("string").str.strip().str.casefold()
city = city.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["city"] = city.str.title()
```

```python
# clean city column

urbannest_df["city"]
```

```python
# clean the region column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA.

region = urbannest_df["region"].astype("string").str.strip().str.casefold()
region = region.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["region"] = region
```

```python
# clean region column

urbannest_df["region"]
```

```python
# clean the customer_segment column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA.

customer_segment=urbannest_df["customer_segment"].astype("string").str.strip().str.casefold()
customer_segment = customer_segment.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["customer_segment"] = customer_segment
```

```python
# clean customer_segment column

urbannest_df["customer_segment"]
```

```python
# clean the delivery_fee column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>", "nan", "free", "0", "0.00", "5", "5.00", "10", and "10.00" with their corresponding numeric values.

delivery_fee=urbannest_df["delivery_fee"].astype("string").str.strip().str.casefold()
urbannest_df["delivery_fee"] = delivery_fee.replace({
    "<na>": pd.NA,
    "nan": pd.NA,
    "free": 0.0,
    "0": 0.0,
    "0.00": 0.0,
    "5": 5.0,
    "5.00": 5.0,
    "10": 10.0,
    "10.00": 10.0,
})
```

```python
# change the delivery_fee column to numeric type, coercing errors to NaN

urbannest_df["delivery_fee"] = pd.to_numeric(urbannest_df["delivery_fee"], errors="coerce")
```

```python
#clean delivery_fee column

urbannest_df["delivery_fee"]
```

```python
# clean the payment_method column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, convert the cleaned payment method names to title case.

payment_method=urbannest_df["payment_method"].astype("string").str.strip().str.casefold()
payment_method = payment_method.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["payment_method"] = payment_method.str.title()
```

```python
#clean payment_method column

urbannest_df["payment_method"]
```

```python
# clean the sales_channel column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, convert the cleaned sales channel names to title case.

sales_channel=urbannest_df["sales_channel"].astype("string").str.strip().str.casefold()
sales_channel = sales_channel.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["sales_channel"] = sales_channel.str.title()
```

```python
# clean sales_channel column

urbannest_df["sales_channel"]
```

```python
# clean the order_status column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, convert the cleaned order status names to title case.

order_status=urbannest_df["order_status"].astype("string").str.strip().str.casefold()
order_status = order_status.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["order_status"] = order_status.str.title()
```

```python
# clean the delivery_status column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, convert the cleaned delivery status names to title case.

delivery_status=urbannest_df["delivery_status"].astype("string").str.strip().str.casefold()
delivery_status = delivery_status.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["delivery_status"] = delivery_status.str.title()
```

```python
# clean the delivery_days column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, replace any range of days (e.g., "3-5") with the lower bound followed by a plus sign (e.g., "3+").

delivery_days = urbannest_df["delivery_days"].astype("string").str.strip().str.casefold()
delivery_days = delivery_days.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["delivery_days"] = delivery_days.str.replace(r"^(\d+)-(\d+)$", r"\1+", regex=True)
```

```python
# clean the customer_rating column by removing leading and trailing spaces, converting to lowercase, and replacing "<na>" and "nan" with pd.NA. Finally, replace any range of ratings (e.g., "3-5") with the lower bound followed by a plus sign (e.g., "3+").

customer_rating = urbannest_df["customer_rating"].astype("string").str.strip().str.casefold()
customer_rating = customer_rating.replace({"<na>": pd.NA, "nan": pd.NA})
urbannest_df["customer_rating"] = customer_rating.str.replace(r"^(\d+)-(\d+)$", r"\1+", regex=True)
```
